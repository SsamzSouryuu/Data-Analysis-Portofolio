## 4. Sales Analysis

Menganalisis nilai penjualan berdasarkan berbagai karakteristik
yang tersedia dalam dataset.

### 4.1 What is the total sales?

Berapa total nilai Sales dari seluruh data transaksi?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

total_sales = data['Sales'].sum()

print(f"Total sales adalah: {total_sales}")
```

### 4.2 What is the average sales per record?

Berapa rata-rata nilai Sales pada setiap record?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

rata_rata = data['Sales'].mean()

print(f"Rata-rata dari Sales adalah: {rata_rata}")
```

### 4.3 Which category generates the highest sales?

Kategori produk mana yang menghasilkan nilai Sales terbesar?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

highest = data.groupby('Category')['Sales'].sum()
nama_kategori = highest.idxmax()
angka_sales = highest.max()
print(f"Kategori terbesar adalah: {nama_kategori}\nDDengan sales sebesar: {angka_sales}")
```

### 4.4 Which sub-category generates the highest sales?

Sub-category mana yang menghasilkan nilai Sales terbesar?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

highest = data.groupby('Sub-Category')['Sales'].sum()
sub_kategori = highest.idxmax()
angka_sales = highest.max()
print(f"Sub-Kategori terbesar adalah: {sub_kategori}\nDengan sales sebesar: {angka_sales}")
```

### 4.5 Which products have the highest sales?

Produk individual mana yang memiliki nilai Sales terbesar?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

highest = data.groupby('Product Name')['Sales'].sum()
nama_product = highest.idxmax()
angka_sales = highest.max()
print(f"Product terbesar adalah: {nama_product}\nDengan sales sebesar: {angka_sales}")
```
