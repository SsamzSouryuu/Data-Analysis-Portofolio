## 4. Sales Analysis

Menganalisis nilai penjualan berdasarkan berbagai karakteristik
yang tersedia dalam dataset.

### 4.1 What is the total sales?

Berapa total nilai Sales dari seluruh data transaksi?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

total_sales = data['Sales'].sum()

print(f"Total sales adalah: {total_sales}")
```

### 4.2 What is the average sales per record?

Berapa rata-rata nilai Sales pada setiap record?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

rata_rata = data['Sales'].mean()

print(f"Rata-rata dari Sales adalah: {rata_rata:.3f}")
```

### 4.3 Which category generates the highest sales?

Kategori produk mana yang menghasilkan nilai Sales terbesar? (Total akumulasi)

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

highest = data.groupby('Category')['Sales'].sum()
nama_kategori = highest.idxmax()
angka_sales = highest.max()
print(f"Kategori terbesar adalah: {nama_kategori}\nDDengan total sales sebesar: {angka_sales}")
```

### 4.4 Which sub-category generates the highest sales?

Sub-category mana yang menghasilkan nilai Sales terbesar? (Total akumulasi)

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

highest = data.groupby('Sub-Category')['Sales'].sum()
sub_kategori = highest.idxmax()
angka_sales = highest.max()
print(f"Sub-Kategori terbesar adalah: {sub_kategori}\nDengan total sales sebesar: {angka_sales}")
```

### 4.5 Which products have the highest sales?

Produk individual mana yang memiliki nilai Sales terbesar? (Total akumulasi)

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

highest = data.groupby('Product Name')['Sales'].sum()
nama_product = highest.idxmax()
angka_sales = highest.max()
print(f"Product terbesar adalah: {nama_product}\nDengan total sales sebesar: {angka_sales}")
```

## 6. Time & Shipping Analysis

Menganalisis pola penjualan berdasarkan waktu dan metode pengiriman.

### 6.1 How does sales change over time?

Bagaimana perubahan Sales berdasarkan waktu?
Apakah terdapat periode dengan Sales yang lebih tinggi atau lebih rendah?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

data['Order Date'] = pd.to_datetime(data['Order Date'], dayfirst=True)

data['Bulan'] = data['Order Date'].dt.to_period('M')
sales_per_bulan = data.groupby('Bulan')['Sales'].sum()

print(sales_per_bulan)
```

### 6.2 Which year has the highest sales?

Tahun mana yang memiliki total Sales terbesar?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

data['Order Date'] = pd.to_datetime(data['Order Date'], dayfirst=True)
data['Tahun'] = data['Order Date'].dt.year

sales_per_tahun = data.groupby('Tahun')['Sales'].sum()

tahun_terbesar = sales_per_tahun.idxmax()
angka_sales = sales_per_tahun.max()

print(f"Tahun dengan sales terbesar: {tahun_terbesar}\nDengan sales sebesar: {angka_sales}")
```

### 6.3 Which month has the highest sales?

Bulan mana yang memiliki total Sales terbesar?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

data['Order Date'] = pd.to_datetime(data['Order Date'], dayfirst=True)
data['Bulan'] = data['Order Date'].dt.month

sales_per_bulan = data.groupby('Bulan')['Sales'].sum()

bulan_terbesar = sales_per_bulan.idxmax()
angka_sales = sales_per_bulan.max()

print(f"Bulan dengan sales terbesar: {bulan_terbesar}\nDengan sales sebesar: {angka_sales}")
```

### 6.4 Which shipping mode is associated with the highest sales?

Metode pengiriman mana yang memiliki total Sales terbesar?

```python
import pandas as pd

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

highest = data.groupby('Ship Mode')['Sales'].sum()
ship = highest.idxmax()
angka_sales = highest.max()
print(f"Metode pengiriman dengan sales terbesar adalah: {ship}\nDengan total sales sebesar: {angka_sales}")
```

## 7. Data Visualization

Memvisualisasikan hasil analisis untuk melihat pola dan perbandingan
dengan lebih mudah.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv ("/kaggle/input/datasets/ismatkhan121/supermarket-sales-dataset/supermarket.csv")

data['Order Date'] = pd.to_datetime(data['Order Date'], dayfirst=True)

data['Bulan'] = data['Order Date'].dt.to_period('M')
sales_per_bulan = data.groupby('Bulan')['Sales'].sum()

sales_per_bulan.plot(kind='line', marker='o', figsize=(12,5))
plt.xlabel('Bulan')
plt.ylabel('Total Sales')
plt.title('Tren Sales per Bulan')
plt.xticks(rotation=45)
plt.show()
```
